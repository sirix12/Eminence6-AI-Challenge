```python
%pip install -q xgboost lightgbm catboost scikit-learn pandas numpy matplotlib seaborn optuna missingno

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import random
import warnings

warnings.filterwarnings('ignore')

SEED = 42
def seed_everything(seed=SEED):
    random.seed(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)
    np.random.seed(seed)
    print(f"Global random seed fixed to: {seed}")

seed_everything(SEED)
```

```python
def save_artifact(obj, filename):
    path = os.path.join(PROJECT_DIR, filename)
    joblib.dump(obj, path)
    print(f"Artifact saved: {path}")

def load_artifact(filename):
    path = os.path.join(PROJECT_DIR, filename)
    if os.path.exists(path):
        print(f"Artifact loaded: {path}")
        return joblib.load(path)
    else:
        print(f"File not found: {path}")
        return None
```
